**## 1. Explain the differences between AI, ML, Deep Learning (DL), and Data Science (DS).**

Artificial Intelligence (AI), Machine Learning (ML), Deep Learning (DL), and Data Science (DS) are interconnected fields of technology, but they have different purposes and applications.

**1. Artificial Intelligence (AI):** AI is a broad field of computer science that focuses on developing machines and systems capable of performing tasks that normally require human intelligence, such as reasoning, problem-solving, decision-making, and understanding language. Example: Virtual assistants like Siri and Alexa.

**2. Machine Learning (ML):** ML is a subset of AI that enables computers to learn from data and improve their performance without being explicitly programmed for every task. It uses algorithms to identify patterns and make predictions. Example: Email spam detection and product recommendations.

**3. Deep Learning (DL):** DL is a subset of ML that uses artificial neural networks with multiple layers to learn complex patterns from large amounts of data. It is particularly useful for image recognition, speech recognition, and natural language processing. Example: Facial recognition and self-driving cars.

**4. Data Science (DS):** DS is an interdisciplinary field that combines statistics, mathematics, programming, and domain knowledge to extract meaningful insights from structured and unstructured data. It involves data collection, cleaning, analysis, visualization, and interpretation. Example: Sales forecasting and customer behaviour analysis.

In Summary, AI is the broader concept, ML is a subset of AI, and DL is a subset of ML. Data Science is a separate but closely related field that uses various techniques, including ML and AI, to extract valuable insights and support data-driven decision-making.

**## 2. What are the types of machine learning? Describe each with one real-world example.**

Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn from data, identify patterns, and make decisions or predictions without being explicitly programmed. Machine Learning is mainly classified into four types:

**1. Supervised Learning:**
In supervised learning, the model is trained using labelled data, where both input data and the correct output are provided. It learns the relationship between them to predict outcomes for new data.
Example: Email spam detection, where the model learns to classify emails as spam or not spam based on previously labelled emails.

**2. Unsupervised Learning:**
In unsupervised learning, the model is trained using unlabelled data. It identifies hidden patterns, relationships, and groups within the dataset without predefined outputs.
Example: Customer segmentation, where a company groups customers based on their purchasing behaviour and preferences.

**3. Semi-Supervised Learning:**
Semi-supervised learning uses a combination of a small amount of labelled data and a large amount of unlabelled data. It improves learning accuracy when obtaining labelled data is expensive or time-consuming.
Example: Image classification, where a model learns from a few labelled images and thousands of unlabelled images.

**4. Reinforcement Learning:**
Reinforcement learning involves an agent learning through interaction with an environment. It receives rewards for correct actions and penalties for incorrect ones, gradually improving its decisions.
Example: A self-driving car learning to navigate roads by receiving rewards for safe driving.

These four types of Machine Learning help solve different real-world problems by enabling systems to learn from data and improve their performance.

**## 3. Define overfitting, underfitting, and the bias-variance tradeoff in machine learning.**

In Machine Learning, overfitting, underfitting, and the bias-variance tradeoff are important concepts that determine how well a model learns from data and performs on unseen data.

**1. Overfitting:**

Overfitting occurs when a model learns the training data too well, including noise and unnecessary details. As a result, it performs very well on training data but poorly on new or unseen data. It usually happens when the model is too complex.
Example: A student memorizes all answers from practice questions but struggles to solve new questions in an examination.

**2. Underfitting:**

Underfitting occurs when a model is too simple to understand the underlying patterns in the training data. It performs poorly on both training and testing data. It generally happens when the model has insufficient complexity or is not trained adequately.
Example: A student studies only a few basic concepts and fails to understand or answer most examination questions.

**3. Bias-Variance Tradeoff:**

The bias-variance tradeoff refers to balancing two types of errors to achieve better model performance.
>**- Bias:** Error caused by overly simple assumptions in a model. High bias leads to underfitting.

>**- Variance:** Error caused by excessive sensitivity to changes in training data. High variance leads to overfitting.

A good Machine Learning model should maintain a balance between bias and variance so that it performs accurately on both training and unseen data, ensuring better generalization.

**## 4. What are outliers in a dataset, and list three common techniques for  handling them.**

Outliers are data points in a dataset that differ significantly from other observations. They have unusually high or low values compared to the general pattern of the data. Outliers may occur due to measurement errors, data entry mistakes, experimental errors, or genuine variations in the dataset. They can affect statistical calculations, distort data distributions, and reduce the accuracy of Machine Learning models.

Three Common Techniques for Handling Outliers:

**1. Removal Method:**
In this technique, outliers are identified and removed from the dataset if they are caused by errors or are irrelevant to the analysis. However, removing genuine observations may result in losing valuable information.
Example: Removing an incorrectly recorded age of 250 years from a customer dataset.

**2. Capping or Winsorization:**
This technique replaces extreme values with predefined upper and lower limits instead of deleting them. It helps reduce the influence of extreme observations while retaining the data.
Example: Replacing unusually high salary values with the 95th percentile value.

**3. Transformation Method:**
Data transformation techniques, such as logarithmic or square root transformation, are used to reduce the impact of extreme values and make the data distribution more balanced.
Example: Applying a logarithmic transformation to highly skewed income data.

Outlier treatment is an important step in data preprocessing. Selecting the appropriate technique helps improve data quality, model performance, and the reliability of analysis.

**##  5. Explain the process of handling missing values and mention one imputation technique for numerical and one for categorical data.**

Missing values refer to absent or unavailable information in a dataset. They may occur due to data entry errors, incomplete surveys, equipment failures, or problems during data collection. Handling missing values is an important step in data preprocessing because they can affect statistical analysis and reduce the accuracy of Machine Learning models.

Process of Handling Missing Values:

**1. Identifying Missing Values:** The first step is to detect missing values using methods such as isnull() or isna() in Python's Pandas library.

**2. Analyzing Missing Values:** Examine the percentage and pattern of missing data to understand their possible causes and determine an appropriate treatment method.

**3. Handling Missing Values:** Missing values can be removed or replaced using imputation techniques, depending on the dataset and the importance of the missing information.

**Imputation Techniques:**

> **1. Numerical Data – Mean Imputation:** This technique replaces missing numerical values with the average of available values in the respective column.
>>Example: Replacing missing employee salary values with the average salary of employees.

>**2. Categorical Data – Mode Imputation:** This technique replaces missing categorical values with the most frequently occurring category in the respective column.

>>Example: Replacing missing city names with the most frequently occurring city.

 Proper handling of missing values improves data quality, reduces potential errors, and helps Machine Learning models produce more accurate and reliable predictions.

**## 8. Write a Python script to:**
>**1. Generate 1000 samples from a normal distribution.**
>**2. Introduce 50 random missing values.**
>**3. Fill missing values with the column mean.**
>**4. Plot a histogram before and after imputation.**


```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Generate 1000 samples from a normal distribution
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)

# Step 2: Create a DataFrame
df = pd.DataFrame(data, columns=['Values'])

# Step 3: Introduce 50 random missing values
missing_indices = np.random.choice(df.index, 50, replace=False)
df.loc[missing_indices, 'Values'] = np.nan

# Step 4: Plot histogram before imputation
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(df['Values'].dropna(), bins=30, edgecolor='black')
plt.title('Before Imputation')
plt.xlabel('Values')
plt.ylabel('Frequency')

# Step 5: Fill missing values with column mean
mean_value = df['Values'].mean()
df['Values'] = df['Values'].fillna(mean_value)

# Step 6: Plot histogram after imputation
plt.subplot(1, 2, 2)
plt.hist(df['Values'], bins=30, edgecolor='black')
plt.title('After Mean Imputation')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

# Print the mean and missing values count
print("Column Mean:", mean_value)
print("Missing Values After Imputation:", df.isnull().sum().values[0])
```

**## 9. Implement Min-Max scaling on the following list of numbers [2, 5, 10, 15, 20] using sklearn.preprocessing MinMaxScaler. Print the scaled array.**


```python

import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Step 1: Create a list of numbers
data = np.array([2, 5, 10, 15, 20]).reshape(-1, 1)

# Step 2: Create MinMaxScaler object
scaler = MinMaxScaler()

# Step 3: Apply Min-Max Scaling
scaled_data = scaler.fit_transform(data)

# Step 4: Print the scaled array
print("Scaled Array:")
print(scaled_data)
```

**Explain the step-by-step data preparation plan you’d follow before training a machine learning model. Include how you’d address missing data, outliers, imbalance, and encoding.**


Before training a Machine Learning model, it is important to prepare the customer transaction dataset properly to improve model accuracy and reliability. I would follow these steps:

**1. Understanding and Exploring the Dataset:**

First, I would examine the dataset using Python libraries such as Pandas and NumPy. I would check the number of rows and columns, data types, duplicate records, missing values, and statistical summaries to understand the dataset.

**2. Handling Missing Values:**

Missing values in the age column can be handled using imputation techniques. Since age is numerical data, I would use median imputation instead of mean imputation because the median is less affected by extreme values. If necessary, I would also create a separate indicator column to identify records where age was missing.

**3. Handling Outliers:**

I would identify outliers in transaction amounts using box plots and the Interquartile Range (IQR) method. Depending on whether the extreme values are genuine or caused by errors, I would either remove incorrect observations or apply capping (Winsorization) to limit extreme values without deleting the records.

**4. Handling Imbalanced Data:**

Since fraudulent transactions are much fewer than non-fraudulent transactions, the dataset is highly imbalanced. I would use techniques such as SMOTE (Synthetic Minority Over-sampling Technique) to generate synthetic minority-class samples. I would also consider class weights to give more importance to fraudulent transactions. Evaluation metrics such as precision, recall, F1-score, and PR-AUC would be preferred over accuracy alone.

**5. Encoding Categorical Variables:**

Categorical variables such as payment method would be converted into numerical values using encoding techniques. I would use One-Hot Encoding for categories such as credit card, debit card, cash, and UPI because these categories have no natural order.

**6. Feature Scaling:**

I would apply Standardization or Min-Max Scaling to numerical features such as age and transaction amount to bring them onto comparable scales.

**7. Splitting the Dataset:**

Finally, I would divide the dataset into training and testing sets using stratified sampling to maintain the fraud-to-non-fraud ratio. All preprocessing steps, including imputation, scaling, and encoding, would be fitted using only the training data. SMOTE would also be applied only to the training data to prevent data leakage.



Proper data preparation is essential for building an effective fraud detection model. By handling missing values, outliers, class imbalance, and categorical variables systematically, we can improve model performance and make more reliable fraud predictions.
